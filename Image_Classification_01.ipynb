```python
import keras
import numpy as np
```

```python
(X_train, y_train),(X_test, y_test) = keras.datasets.mnist.load_data(path="mnist.npz")
```

```python
X_train[0]
```

```python
y_train[0]
```

```python
np.std(X_train) # Check SD, if > 10 you will need to normalise
```

Image classification, is normal to normalise as SD is normally high

```python
X_train = X_train/255. #255 because of max RGB value
```

```python
X_test = X_test/255
```

```python
np.std(X_train) # check is SD is between 0-10
```

```python
from keras.models import Sequential
from keras.layers import Dense, Flatten, Activation
```

```python
model = Sequential()
model.add(keras.layers.Flatten(input_shape=[28,28])) #Flatten 3d array to 1d
model.add(keras.layers.Dense(10, activation = "softmax")) # Determine how many layers your Neural Network has
```

Activation has 6 formulas, often using "relu" and "softmax"

```python
model.compile(loss="sparse_categorical_crossentropy",
              optimizer="adam",
              metrics=["accuracy"])
```

```python
model.fit(X_train, y_train, epochs=5) # fit model
```

epochs = how many times you want to run the code, aka to train. default is 1.

```python
model.evaluate(X_test, y_test) # Evaluate model
```

loss = 0.267, closer to 0 the better
accuracy = 0.925, closer to 1 the better

```python
predicted = model.predict(X_test)
```
